```python
#int
revenue=50
expenses=10
profit=revenue-expenses
print(f"profit is:{profit}")
type(profit)
#float
margin=profit*100/revenue
print(f"Margin is : {margin:.2f}")#here .2f has to be divided with a :
```

```python
#lists
rev=[50,60,70]
exp=[10,20,30]
rev[0:3]#[50, 60, 70]
rev[1:3]#[60, 70]
exp[-2]#20
#for loop
for i in rev:
    print(i, end=",")#to print in the same line
    #print(i) to print one by one
print()#a new line
for i in range (len(rev)):
    profit=rev[i]-exp[i]
    margin=profit*100/rev[i]
    print(f"margin is :{margin:.2f}")
```
