```python
#lists
rev=[50,60,70]
exp=[10,20,30]
rev[0:3]#[50, 60, 70]
rev[1:3]#[60, 70]
exp[-2]#20
#for loop
for i in rev:
    print(i, end=",")#to print in the same line
    #print(i) to print one by one
print()#a new line
for i in range (len(rev)):
    profit=rev[i]-exp[i]
    margin=profit*100/rev[i]
    print(f"margin is :{margin:.2f}")
    
for i in range (len(rev)):
    profit=rev[i]-exp[i]
    margin=profit*100/rev[i]
    print(f"margin Q{i+1} :{margin:.2f}")
margins=[]
for i in range (len(rev)):
    profit=rev[i]-exp[i]
    margin=profit*100/rev[i]
    margins.append(margin)
print(margins)
```

```python
fins={'Q1':{"rev":50,"exp":10},'Q2':{"rev":60,"exp":20},'Q3':{"rev":70,"exp":30}}
#fins
for qtr,data in fins.items():
    profit=data['rev']-data['exp']
    #data['margin']=profit*100/data['rev']
    margin=profit*100/data['rev']
    print(f"{qtr}: {margin:.2f}%")
```
